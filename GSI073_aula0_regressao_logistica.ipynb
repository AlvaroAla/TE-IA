# GSI073 - Tópicos Especiais de Inteligência Artificial

## Definição dos dados

```python
import torch; import sklearn

# 1. Carregar dados
iris = sklearn.datasets.load_iris()
X = iris.data        # 4 features: sépalas e pétalas
y = (iris.target == 1).astype(float)  # 1 se Versicolor, 0 caso contrário

# 2. Preparar dados para pytorch
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32).view(-1, 1)
```

## Definição do modelo e treinamento

```python
# 3. Definir modelo: regressão logística
modelo = torch.nn.Linear(4, 1)  # 4 features → 1 saída (probabilidade de ser Versicolor)

# 4. Definir função de perda e algoritmo de otimização
funcao_perda = torch.nn.BCEWithLogitsLoss()  # combinação de sigmoid + BCE
optimizer = torch.optim.SGD(modelo.parameters(), lr=0.1)
```

## Execução do treinamento

```python
# 5. Treino
for epoch in range(1000):
    optimizer.zero_grad() # reseta gradiente senão acumula
    outputs = modelo(X)
    loss = funcao_perda(outputs, y)
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 10 == 0:
        print(f"Época [{epoch+1}/100], Loss: {loss.item():.4f}")
```

##Resposta do Exercicio do slide

```python
# --- Avaliação do modelo ---

import torch
from sklearn.metrics import accuracy_score, confusion_matrix

# 1. Obter as predições (logits → sigmoid p/ probabilidade)
with torch.no_grad():
    logits = modelo(X)
    probs = torch.sigmoid(logits)
    preds = (probs >= 0.5).float()

# 2. Calcular acurácia
acc = accuracy_score(y.numpy(), preds.numpy())

# 3. Matriz de confusão
cm = confusion_matrix(y.numpy(), preds.numpy())

print("Acurácia do modelo:", acc)
print("\nMatriz de Confusão:\n", cm)

# Mostrar algumas previsões
for i in range(5):
    print(f"Amostra {i}: Prob={probs[i].item():.4f}, Pred={preds[i].item()}, Real={y[i].item()}")
```

```python
# --- Inspecionar o nn.Linear ---

print("=== Estrutura do modelo ===")
print(modelo)

print("\n=== Pesos (W) ===")
print(modelo.weight)
print("Formato dos pesos:", modelo.weight.shape)

print("\n=== Bias (b) ===")
print(modelo.bias)
print("Formato do bias:", modelo.bias.shape)

# Número de parâmetros treináveis
total_params = sum(p.numel() for p in modelo.parameters())
print("\nTotal de parâmetros treináveis:", total_params)
```
